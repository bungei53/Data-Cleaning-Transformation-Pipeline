# Chardonnay Wine Data Exploration & Cleaning

This notebook demonstrates exploratory data analysis (EDA) and data cleaning on a dataset of scraped Chardonnay wine information.

## Project Steps Completed:
1. **Data Loading**: Loaded the CSV dataset (`all_scraped_wine_info_chardonnay.csv`) containing wine reviews and specifications.
2. **Feature Inspection**: Examined the active columns, shape, and overall structure of the dataset.
3. **Data Cleaning & Conversion**:
   - Cleaned the `Alcohol` percentage column by stripping characters and converting it into a numeric float representation (`alcohol_percent`).
   - Isolated and inspected high-percentage outliers (e.g., records with alcohol levels flagged over 50%).
4. **Feature Engineering**: Extracted clean formatting from the `Date Published` feature for subsequent analysis.

```python
# Clean the 'Alcohol' column by removing the percentage symbol, stripping whitespace,
# and converting the values to floating-point numbers for numeric analysis.
df['alcohol_percent'] = (df['Alcohol'].str.replace('%', "", regex=False)
                         .str.strip()
                         .astype(float))
```

```python
df.loc[(df['alcohol_percent']>=50) & (df['alcohol_percent']<=120),['Rating','Vintage','alcohol_percent','Subregion','Reviewer','Category']]
```

```python
df['Date Published'].dtype
```

```python
# Strip out the slashes from 'Date Published' to format the date string
# into a raw sequence of numbers for parsing or extraction purposes.
df['data'] = df['Date Published'].str.replace('/', '', regex=False)
```

```python
df.columns
```

```python
# Convert 'Date Published' to a proper datetime format
df['Date Published'] = pd.to_datetime(df['Date Published'], errors='coerce')

# Display the updated data type and a preview of the column
print("New data type:", df['Date Published'].dtype)
display(df['Date Published'].head())
```

```python
# Display summary statistics for 'Price' and 'Rating'
summary_stats = df[['Price', 'Rating']].describe()
display(summary_stats)
```

```python
# Clean the 'Price' column by removing '$' and commas, then convert to numeric
df['Price'] = df['Price'].astype(str).str.replace('$', '', regex=False).str.replace(',', '', regex=False)
df['Price'] = pd.to_numeric(df['Price'], errors='coerce')

# Display updated summary statistics for both Price and Rating
display(df[['Price', 'Rating']].describe())
```
